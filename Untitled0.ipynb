Paso 5: Limpiar los datos

Elimina espacios de mas al inicio y final de cada texto. ejemplo " MODA " se converte en "MODA". esto es necesario por que si tiene espacios, esta palabras no son iguales.

```python
# Seleccionar columnas de texto
columnas_texto =df.select_dtypes(include='object').columns
# limpiar cada columna
for columna in columnas_texto:
  df[columna] = df[columna].str.strip()
#mostrar reslitado de la limpieza que fue completada
print('limpieza completada')
```

punto 6 Distribucion por sector


Que hace? Cuenta cuantos comercion hay en cada sector y calcula el procentaje.

```python
#value_conts() cuenta las veces que aparece cada valor
# normalize = True  lo converte en porcentaje
tabla_sector =pd.DataFrame({'Cantidad' : df['Categoría / Sector'].value_counts(),'Porcentaje (%)' : (df['Categoría / Sector'].value_counts(normalize=True)*100).round(2)})

print("Distribucion por sector:")
display(tabla_sector)
```

# **7 Tablas cruzadas**

# que hace? Muestra la relacion entre dos varibales. Ejemplo Que sectores estan dispuestos a pagar mas ?

```python
# pd.crosstab() Crea  una tabla que cruza  dos variables
# las filas son sectores ,  las columnas son rangos de precio
crosstab =pd.crosstab(df['Categoría / Sector'],df['P15_PrecioSaaS'],margins=True)
print(" tabla cruzada vr precio por sector")
display(crosstab)
```

# 8 KPIs indicadores clave

¿Qué hace? Calcular porcentajes importantes para el negocio. Ejemplo: CUntaos clientes pago 1-clic?

```python
# Kpi 1 : % que consedera **ALto** en impacto de pago 1-CLic
#str.contains() busca texto , .mena() calcula promedio, * 100 converte a %
kpi1 =df['P2_Impacto1Clic'].str.contains('Alto|Crítico', case=False, na=False).mean()*100
# KPI 2: % Con problemas de overselling frecuente
kpi2= df['P7_FrecOverselling'].str.contains('Frecuentemente|Diariamente', case=False, na=False).mean()*100
print(f"KPI 1: Impacto 1_clic alto|Crítico {kpi1:.2f}%")
print(f"KPI 2:Overselling Frecuente|Díario {kpi2:.2f}%")
```

# 9 Mapeo de texto a Numeros

¿Qué hace? CONVERTIR respuesta de texto a número para poder hacer cálculo. Ejemplo: '3 - 4 canales' se convierte en 3.5

```python
# Diccionario de conversiones
mapa_canales ={'1 canal':1.0, '2 canales':2.0, '3-4 canles':3.5, '5 o mas canaes':5.0}
# aplicar las conversiones
df['NumCanales_Num'] = df['P6_NumCanales'].map(mapa_canales)
#Calcular estadisticas
print(f"Promedio de calanes: {df['NumCanales_Num'].mean():.2f}")
print(f"Mediana de calanes: {df['NumCanales_Num'].median():.2f}")
```

# 10 Método de pago por sector
¿Que hace?
para cara sector,  encuentra cual es el método de pago más elegido.
1. función para obtener la moda de forma segura
2. La "moda" es el valor que más se repite

```python
print("Método de pago mas usado por sector:")
def moda_segura(x):
  m=x.mode()
  return m.iloc[0] if not m.empty else 'sin datos'

# agrupar pro sector y encontrar el método  mas usado
pago_por_sector= df.groupby('Categoría / Sector')['P3_PagoPref'].agg(moda_segura)
print("Método de pago mas usado por sector:")
print(pago_por_sector)
```

```python
print("Método de pago mas usado por sector:")
def moda_segura(x):
  m=x.mode()
  return m.iloc[0] if not m.empty else 'sin datos'

# agrupar pro sector y encontrar el método  mas usado
pago_por_sector= df.groupby('Categoría / Sector')['P3_PagoPref'].agg(moda_segura)
print("Método de pago mas usado por sector:")
print(pago_por_sector)
```

# 11 Disposición a pagar comisión

¿Qué hace? Muestra qué porcentaje de comercion aceptaría pagar una comisión extra del 0.5%.

```python
# value_conunt(normalize =True) calcula proporciones
comision_dist=df['P5_Comision1Clic'].value_counts(normalize=True)*100
print("¿Cuantos aceptaria pagar comisión del 0.5%")
for respuesta, porcentaje in comision_dist.items():
  print(f"{respuesta}: {porcentaje:.1f}%")
```

# 12.1  graficas de barras horinzontales
¿Qué muestra? Se usan barras horizontales porque los  nombres son largos.

```python
plt.figure(figsize=(10,5))
# contar cuantos eligieron cada funncionalidad
upsell_counts= df['P13_FuncionalidadUpsell'].value_counts()
# crear barars horizontales
sns.barplot(x=upsell_counts.values,y=upsell_counts.index,palette='viridis')
# colocacion de tidulos para el diagrama
plt.title('¿Que funcionalesdes del UPsell son más demandadas?')
plt.xlabel('Número de comercio')
plt.ylabel('Funcionalidad')
plt.tight_layout()
plt.show()
```

12.2 Gráfico Circular (Pie Chart)

¿Que muestra? ¿Cuaánto están dispuestos a pagar los comercions? Se usa circular por que muestra parte de un todo ( porcentajes).

```python
plt.figure(figsize=(7,7))
# Calcular el porcentaje de cada
#COntar respuestas de precio
p_counts= df['P15_PrecioSaaS'].value_counts()
# COntar respuestas de precio
p_counts= df['P15_PrecioSaaS'].value_counts()
# Colores y explosion
colors=['#1f77b4','#ff7f0e','#2ca02c','#d62728','#9467bd','#8c564b']
colors=colors[:len(p_counts)]
explode =[0.08 if i == 0 else 0 for i in range(len(p_counts))]
# Crear el gráfico circular
plt.pie(p_counts,labels=p_counts.index, autopct='%1.1f%%',startangle=140,colors=colors, explode=explode)
plt.title('¿Cuanto están dispuestos a pagar mesualmente?')
plt.tight_layout()
plt.show()

```
